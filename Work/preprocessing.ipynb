```python
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
from torch.autograd import grad
from sklearn.preprocessing import MinMaxScaler
```

```python
# Separate features and FLAG
X = df.drop("FLAG", axis=1)
y = df["FLAG"]

# Minority class (theft = 1)
X_minority = X[y == 1]

# Normalize features
scaler = MinMaxScaler()
X_minority_scaled = scaler.fit_transform(X_minority)

X_minority_tensor = torch.tensor(X_minority_scaled, dtype=torch.float32)
```

```python
class Generator(nn.Module):
    def __init__(self, noise_dim, output_dim):
        super().__init__()
        self.model = nn.Sequential(
            nn.Linear(noise_dim, 128),
            nn.ReLU(),
            nn.Linear(128, 256),
            nn.ReLU(),
            nn.Linear(256, output_dim)
        )

    def forward(self, z):
        return self.model(z)
```

```python
class Critic(nn.Module):
    def __init__(self, input_dim):
        super().__init__()
        self.model = nn.Sequential(
            nn.Linear(input_dim, 256),
            nn.LeakyReLU(0.2),
            nn.Linear(256, 128),
            nn.LeakyReLU(0.2),
            nn.Linear(128, 1)
        )

    def forward(self, x):
        return self.model(x)
```

```python
def gradient_penalty(critic, real, fake):
    batch_size = real.size(0)
    epsilon = torch.rand(batch_size, 1).to(real.device)
    epsilon = epsilon.expand_as(real)

    interpolated = epsilon * real + (1 - epsilon) * fake
    interpolated.requires_grad_(True)

    mixed_scores = critic(interpolated)

    grad_outputs = torch.ones_like(mixed_scores)
    gradients = grad(
        outputs=mixed_scores,
        inputs=interpolated,
        grad_outputs=grad_outputs,
        create_graph=True,
        retain_graph=True,
        only_inputs=True
    )[0]

    gradients = gradients.view(gradients.size(0), -1)
    gp = ((gradients.norm(2, dim=1) - 1) ** 2).mean()
    return gp
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"

latent_dim = 32
lr = 1e-4
epochs = 800
batch_size = 128
lambda_gp = 10
n_critic = 5

G = Generator(latent_dim, X_minority.shape[1]).to(device)
D = Critic(X_minority.shape[1]).to(device)

opt_G = optim.Adam(G.parameters(), lr=lr, betas=(0.5, 0.9))
opt_D = optim.Adam(D.parameters(), lr=lr, betas=(0.5, 0.9))

data_loader = torch.utils.data.DataLoader(
    X_minority_tensor, batch_size=batch_size, shuffle=True
)

for epoch in range(epochs):
    for real in data_loader:
        real = real.to(device)

        # Train Critic
        for _ in range(n_critic):
            z = torch.randn(real.size(0), latent_dim).to(device)
            fake = G(z)

            real_score = D(real)
            fake_score = D(fake.detach())

            gp = gradient_penalty(D, real, fake)
            d_loss = fake_score.mean() - real_score.mean() + lambda_gp * gp

            opt_D.zero_grad()
            d_loss.backward()
            opt_D.step()

        # Train Generator
        z = torch.randn(real.size(0), latent_dim).to(device)
        fake = G(z)
        g_loss = -D(fake).mean()

        opt_G.zero_grad()
        g_loss.backward()
        opt_G.step()

    if epoch % 200 == 0:
        print(f"Epoch [{epoch}/{epochs}] | D Loss: {d_loss.item():.4f} | G Loss: {g_loss.item():.4f}")
```

```python
import numpy as np
# 5. Trend Slope
def calculate_trend(row):
    x = np.arange(len(row))
    y = row.values
    if np.std(y) == 0:
        return 0
    slope = np.polyfit(x, y, 1)[0]
    return slope
df['trend_slope'] = df[date_columns].apply(calculate_trend, axis=1)
```

```python
df.head()
```

```python
# 6. Max Drop Percentage
pct_changes = df[date_columns].pct_change(axis=1)
df['max_drop_pct'] = pct_changes.min(axis=1).abs()
```

```python
# 7. Sudden Drop Count (drops > 50% of mean)
drops = df[date_columns].diff(axis=1)
df['sudden_drop_count'] = (drops < -df['mean_consumption'].values[:, None] * 0.5).sum(axis=1)
```

```python
# 8. Minimum Consumption
df['min_consumption'] = df[date_columns].min(axis=1)
```

```python
df['max_consumption'] = df[date_columns].max(axis=1)
```

```python
df['median_consumption'] = df[date_columns].median(axis=1)
```

```python
df['num_below_mean'] = df[date_columns].lt(df['mean_consumption'], axis=0).sum(axis=1)
```

```python
df['skewness'] = df[date_columns].skew(axis=1)
```

```python
df['iqr'] = df[date_columns].quantile(0.75, axis=1) - df[date_columns].quantile(0.25, axis=1)
```
